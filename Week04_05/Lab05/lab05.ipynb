```python
from collections import deque
```

```python
graph = {
    'A': ['B','C'],
    'B': ['D','E'],
    'C': ['F'],
    'D': [],
    'E': ['F'],
    'F': []
}
```

```python
def bfs_search(graph,start, target):
    visited=set() # if a node is visited come here
    queue=deque() # 

    visited.add(start) #
    queue.append(start)# 

    while queue: # how many elements in queue it will run
        node=queue.popleft() # the first node come here in this case is 'A'

        if node== target: # match
            print("Target found: ",node) # 
            return True # means terminat here
        for neighbor in graph[node]:# neighber of visited element here at node will come 'A' at neighber is B, C
            if neighbor not in visited: # 
                visited.add(neighbor) # 
                queue.append(neighbor)
    print("Target not found") # 
    return False # 
```

```python
bfs_search(graph,'A', 'F')
```

```python
import networkx as nx
import matplotlib.pyplot as plt
```

```python
adj_list = {
    'A': [('B',1),('F',2)],
    'B': [('C',3)],
    'C': [('D',4),('E',5)],
    'D': [],
    'E': [('D',6)],
    'F': [('E',7), ('B',8)]
}
```

```python
G=nx.DiGraph()
```

```python
for node in adj_list:
    for neighbor, weight in adj_list[node]:
        G.add_edge(node,neighbor,weight=weight) # here weight means edge weight like 2,3,4 etc, neighber alphabets
pos=nx.spring_layout(G) # the structure will be in a layout, proper position
nx.draw(
    G,
    pos,
    with_labels=True,
    node_size=500
)

edge_labels=nx.get_edge_attributes(G,'weight')
nx.draw_networkx_edge_labels(
    G,
    pos,
    edge_labels=edge_labels
)
plt.title("Dijskta Graph")
plt.show()
```
