## 3. Bar Chart
- Bivariate Analysis
- Numerical vs Categorical
- Use case - Aggregate analysis of groups

```python
children=[20,30,10,50,35,25]
colors=['red','blue','green','orange','black','pink']
plt.bar(colors,children)
plt.title("Children favourite colors")
plt.xlabel("Favourite Colors")
plt.ylabel("Childrens Values")
```

```python
children=[20,30,10,50,35,25]
colors=['red','blue','green','orange','black','pink']
plt.bar(colors,children,color=colors)
plt.title("Children favourite colors")
plt.xlabel("Favourite Colors")
plt.ylabel("Childrens Values")
```

```python
children=[20,30,10,50,35,25]
colors=['red','blue','green','orange','black','pink']
plt.barh(colors,children,color=colors)
```

```python
plt.bar(df['batsman'],df['2015'],width=0.5)# good but plot at same location
```

```python
plt.bar(np.arange(df.shape[0])-0.2,df['2015'],width=0.2) # here -0.2 means that before the exact location i.e 0,1,2..
plt.bar(np.arange(df.shape[0]),df['2016'],width=0.2) # at exact location
plt.bar(np.arange(df.shape[0])+0.2,df['2017'],width=0.2)# here +0.2 means that after the exact location i.e 0,1,2..
```

```python
plt.bar(np.arange(df.shape[0])-0.2,df['2015'],width=0.2) 
plt.bar(np.arange(df.shape[0]),df['2016'],width=0.2) 
plt.bar(np.arange(df.shape[0])+0.2,df['2017'],width=0.2)
# used rotation if the names are too large
plt.xticks(np.arange(df.shape[0]),df['batsman'],rotation=45) # replaced 1,2,3 with its name
```

```python
plt.bar(np.arange(df.shape[0])-0.2,df['2015'],width=0.2) 
plt.bar(np.arange(df.shape[0]),df['2016'],width=0.2) 
plt.bar(np.arange(df.shape[0])+0.2,df['2017'],width=0.2)

plt.xticks(np.arange(df.shape[0]),df['batsman'])

plt.show() # if you do not want the extra info
```

```python
plt.bar(np.arange(df.shape[0])-0.2,df['2015'],width=0.2) 
plt.bar(np.arange(df.shape[0]),df['2016'],width=0.2) 
plt.bar(np.arange(df.shape[0])+0.2,df['2017'],width=0.2)

plt.xticks(np.arange(df.shape[0]),df['batsman'].tolist(),) # removed the error by Converting the Series to a list
plt.show()
```

```python
# Stack Bar chart
plt.bar(df['batsman'],df['2017'],label='2017')
plt.bar(df['batsman'],df['2016'],bottom=df['2017'],label='2016') 
plt.bar(df['batsman'],df['2015'],bottom=(df['2017']+df['2016']),label='2015')
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(df.shape[0])

# Bars
b1 = plt.bar(x, df['2017'], label='2017')
b2 = plt.bar(x, df['2016'], bottom=df['2017'], label='2016')
b3 = plt.bar(x, df['2015'], bottom=df['2017']+df['2016'], label='2015')

# Add values on each segment
for i in range(len(df)):
    # 2017 (bottom)
    plt.text(x[i], df['2017'][i]/2, str(df['2017'][i]), ha='center', color='white')

    # 2016 (middle)
    plt.text(x[i], df['2017'][i] + df['2016'][i]/2, str(df['2016'][i]), ha='center', color='white')

    # 2015 (top)
    plt.text(x[i], df['2017'][i] + df['2016'][i] + df['2015'][i]/2, str(df['2015'][i]), ha='center', color='white')

# Labels
plt.xticks(x, df['batsman'], rotation=45)
plt.legend()
plt.show()
```
