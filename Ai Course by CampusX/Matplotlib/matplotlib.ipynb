```python
tips=sns.load_dataset('tips') # built in dataset
tips.sample(10)
```

```python
plt.scatter(tips['total_bill'],tips['tip'])
plt.title('bill vs tip')# tip means the extra reward given to servant for the good service
plt.xlabel("Total Bill")
plt.ylabel("Tip")
```

```python
plt.scatter(tips['total_bill'],tips['tip'],s=tips['size'])# means the more the customer will big the dot
plt.title('bill vs tip')# tip means the extra reward given to servant for the good service
plt.xlabel("Total Bill")
plt.ylabel("Tip")
```

```python
plt.scatter(tips['total_bill'],tips['tip'],s=tips['size']*20)# it means if 2 customer consider as 40 for visualization we made it easy
plt.title('bill vs tip')# tip means the extra reward given to servant for the good service
plt.xlabel("Total Bill")
plt.ylabel("Tip")
```
