```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
iris=pd.read_csv('iris.csv')
```

```python
iris.sample(5)
```

```python
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'])
```

```python
iris['Species']=iris['Species'].replace({'Iris-setosa':0,'Iris-versicolor':1,'Iris-virginica':2}) #replaced with the 0,1,2
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'])# c means color repaced with 0,1,2
```

```python
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'],cmap='jet')# can search for it on matplotlib.org
```

```python
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'],cmap='jet',alpha=0.3)
```

```python
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'],cmap='jet')
plt.colorbar()
```

```python
scatter=plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'],cmap='jet')
plt.colorbar()
# Create legend
handles, _ = scatter.legend_elements()

labels = ['Iris-setosa', 'Iris-versicolor', 'Iris-virginica']
plt.legend(handles, labels, title="Species")

plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
plt.title('Iris Scatter Plot')

plt.show()
```
