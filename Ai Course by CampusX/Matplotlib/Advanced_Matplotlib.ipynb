```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
iris=pd.read_csv('iris.csv')
```

```python
iris.sample(5)
```

```python
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'])
```

```python
iris['Species']=iris['Species'].replace({'Iris-setosa':0,'Iris-versicolor':1,'Iris-virginica':2}) #replaced with the 0,1,2
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'])# c means color repaced with 0,1,2
```

```python
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'],cmap='jet')# can search for it on matplotlib.org
```

```python
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'],cmap='jet',alpha=0.3)
```

```python
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'],cmap='jet')
plt.colorbar()
```

```python
plt.figure(figsize=(9,5))
scatter=plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'],cmap='jet')
plt.colorbar()
# Create legend
handles, _ = scatter.legend_elements()

labels = ['Iris-setosa', 'Iris-versicolor', 'Iris-virginica']
plt.legend(handles, labels, title="Species")

plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
plt.title('Iris Scatter Plot')

plt.show()
```

```python
df=pd.read_csv('batter.csv')
df
```

```python
sample_df=df.head(100).sample(25,random_state=5)
sample_df
```

```python
sample_df.shape
```

```python
plt.scatter(sample_df['avg'],sample_df['strike_rate'])
```

```python
plt.scatter(sample_df['avg'],sample_df['strike_rate'])
for i in range(sample_df.shape[0]):
    plt.text(sample_df['avg'].values[i],sample_df['strike_rate'].values[i],sample_df['batter'].values[i])
```

```python
plt.figure(figsize=(18,9))
plt.scatter(sample_df['avg'],sample_df['strike_rate'],s=sample_df['runs'])
for i in range(sample_df.shape[0]):
    plt.text(sample_df['avg'].values[i],sample_df['strike_rate'].values[i],sample_df['batter'].values[i])
```

```python
plt.figure(figsize=(18,9))
plt.scatter(sample_df['avg'],sample_df['strike_rate'],s=sample_df['runs'])
for i in range(sample_df.shape[0]):
    plt.text(sample_df['avg'].values[i],sample_df['strike_rate'].values[i],sample_df['batter'].values[i],fontdict={'size':12,'color':'brown'})
plt.title("Strike rate vs Average",fontdict={'size':50})
plt.xlabel("Average",fontdict={'size':30})
plt.ylabel('Strike Rate',fontdict={'size':30})
```

```python
plt.figure(figsize=(18,9))
plt.scatter(sample_df['avg'],sample_df['strike_rate'],s=sample_df['runs'])
plt.axhline(130,color='red') # easy to guess the batsman with strike rate above 130 and avg>30
plt.axvline(30,color='red')
plt.axhline(140,color='green')
for i in range(sample_df.shape[0]):
    plt.text(sample_df['avg'].values[i],sample_df['strike_rate'].values[i],sample_df['batter'].values[i],fontdict={'size':12,'color':'brown'})
plt.title("Strike rate vs Average",fontdict={'size':50})
plt.xlabel("Average",fontdict={'size':30})
plt.ylabel('Strike Rate',fontdict={'size':30})
```

## Sub Plots


```python
fig,ax=plt.subplots()
ax.scatter(df['avg'],df['strike_rate'],marker='+',color='red')
ax.set_title("average vs strike rate")
ax.set_ylabel('strike rate')
ax.set_xlabel('average')
```

```python
fig,ax=plt.subplots(nrows=2,ncols=1) # means 2 rows and 1 column
ax[0].scatter(df['avg'],df['strike_rate'])
ax[0].set_title("average vs strike rate")
ax[0].set_ylabel('strike rate')
ax[0].set_xlabel('average')

ax[1].scatter(df['avg'],df['runs'])
ax[1].set_title("average vs strike rate")
ax[1].set_ylabel('strike rate')
ax[1].set_xlabel('average')
```

```python
fig,ax=plt.subplots(nrows=2,ncols=1,sharex=True,figsize=(15,9))
ax[0].scatter(df['avg'],df['strike_rate'])
ax[0].set_title("average vs strike rate")
ax[0].set_ylabel('strike rate')

ax[1].scatter(df['avg'],df['runs'],color='red')
ax[1].set_title("average vs runs")
ax[1].set_ylabel('runs')
ax[1].set_xlabel('average')
fig.show() # remove the technical info
```

```python
fig,ax=plt.subplots(nrows=2,ncols=2,figsize=(15,9))
ax[0,0].scatter(df['avg'],df['strike_rate'],color='green')
ax[0,1].scatter(df['avg'],df['runs'],color='red')
ax[1,0].hist(df['avg'],color='pink')
ax[1,1].hist(df['runs'],color='orange')
fig.show()
```

```python
fig,ax=plt.subplots(nrows=2,ncols=2,figsize=(15,9))
ax[0,0].scatter(df['avg'],df['strike_rate'],color='green')
ax[0,1].scatter(df['avg'],df['runs'],color='red')
ax[1,0].hist(df['avg'],color='pink')
ax[1,1].hist(df['runs'],color='orange')

ax[0,0].set_title("average vs strike rate")
ax[0,0].set_ylabel('strike rate')
ax[0,0].set_xlabel('average')


ax[0,1].set_title("average vs runs")
ax[0,1].set_ylabel('runs')
ax[0,1].set_xlabel('average')


ax[1,0].set_title("average")
ax[1,0].set_xlabel('average')


ax[1,1].set_title("strike rate")
ax[1,1].set_xlabel('strike rate')


fig.show() # remove the technical info
```

### Another way to plot a sub graph

```python
fig=plt.figure()
ax1=fig.add_subplot(2,2,1) # 2 rows, 2 columns, 1 graph 
ax1.scatter(df['avg'],df['strike_rate'],color='green')

ax1=fig.add_subplot(2,2,2)
ax1.scatter(df['avg'],df['runs'],color='red')

ax1=fig.add_subplot(2,2,3)
ax1.hist(df['avg'],color='pink')

ax1=fig.add_subplot(2,2,4)
ax1.hist(df['runs'],color='pink')
```

## 3D line Surface


```python
x=np.linspace(-10,10,100)
y=np.linspace(-10,10,100)
```

```python
xx,yy=np.meshgrid(x,y)
```

```python
xx.shape
```

```python
z=xx*2+yy**2
```

```python
z.shape
```

```python
fig=plt.figure(figsize=(10,8))
ax=fig.add_subplot(projection='3d')
ax.plot_surface(xx,yy,z,cmap='viridis')
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-10, 10, 100)
y = np.linspace(-10, 10, 100)

xx, yy = np.meshgrid(x, y)

z = xx*2 + yy**2   # FIXED

fig = plt.figure(figsize=(10,8))
ax = fig.add_subplot(projection='3d')

ax.plot_surface(xx, yy, z, cmap='viridis')

plt.show()
```
