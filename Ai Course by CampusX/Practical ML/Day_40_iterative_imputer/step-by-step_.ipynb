```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression

# Initial mean imputation
df_current = df0.copy()

tol = 0.001
max_iter = 100

for iteration in range(max_iter):

    df_new = df_current.copy()

    # -------------------------
    # Impute R&D Spend
    # -------------------------
    df_new.iloc[1,0] = np.nan

    X = df_new.iloc[[0,2,3,4],[1,2]]
    y = df_new.iloc[[0,2,3,4],0]

    lr = LinearRegression()
    lr.fit(X,y)

    pred = lr.predict(df_new.iloc[1,[1,2]].values.reshape(1,-1))[0]
    df_new.iloc[1,0] = pred

    # -------------------------
    # Impute Administration
    # -------------------------
    df_new.iloc[3,1] = np.nan

    X = df_new.iloc[[0,1,2,4],[0,2]]
    y = df_new.iloc[[0,1,2,4],1]

    lr = LinearRegression()
    lr.fit(X,y)

    pred = lr.predict(df_new.iloc[3,[0,2]].values.reshape(1,-1))[0]
    df_new.iloc[3,1] = pred

    # -------------------------
    # Impute Marketing Spend
    # -------------------------
    df_new.iloc[4,2] = np.nan

    X = df_new.iloc[0:4,0:2]
    y = df_new.iloc[0:4,2]

    lr = LinearRegression()
    lr.fit(X,y)

    pred = lr.predict(df_new.iloc[4,0:2].values.reshape(1,-1))[0]
    df_new.iloc[4,2] = pred

    # -------------------------
    # Check convergence
    # -------------------------
    diff = abs(df_new - df_current)

    print(f"\nIteration {iteration+1}")
    print(df_new)
    print("\nDifference")
    print(diff)

    if diff.max().max() < tol:
        print(f"\nConverged after {iteration+1} iterations")
        break

    df_current = df_new.copy()

print("\nFinal Imputed DataFrame")
print(df_current)
```
