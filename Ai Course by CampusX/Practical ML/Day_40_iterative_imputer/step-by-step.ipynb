```python
import pandas as pd
import numpy as np

from sklearn.linear_model import LinearRegression
```

```python
df=np.round(pd.read_csv('50_Startups.csv')[['R&D Spend', 'Administration', 'Marketing Spend', 'Profit']]/10000)
np.random.seed(9) #every time you run the code, the same 5 rows are selected
df=df.sample(5)
df
```

```python
df=df.iloc[:,:-1]
df
```

```python
# introducing missing values
df.iloc[1,0]=np.nan
df.iloc[3,1]=np.nan
df.iloc[4,2]=np.nan
df
```

```python
# df0 to fill with mean values
df0=df.copy()
df0['R&D Spend'].fillna(df0['R&D Spend'].mean(),inplace=True)
df0['Administration'].fillna(df0['Administration'].mean(),inplace=True)
df0['Marketing Spend'].fillna(df0['Marketing Spend'].mean(),inplace=True)
df0
```
