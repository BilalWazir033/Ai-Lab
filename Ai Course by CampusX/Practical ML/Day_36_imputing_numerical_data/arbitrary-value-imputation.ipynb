```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
```

```python
df = pd.read_csv('titanic_toy.csv')
```

```python
df.head()
```

```python
df.isnull().mean()
```

```python
X=df.drop(columns=['Survived'])
y=df['Survived']
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
X_train['Age_99']=X_train['Age'].fillna(99)
X_train['Age_minus1']=X_train['Age'].fillna(-1)

X_train['Fare_999']=X_train['Fare'].fillna(999)
X_train['Fare_minus1']=X_train['Fare'].fillna(-1)
```

```python
print('Original Age Variable Variance: ',X_train['Age'].var())
print('Age Variance after 99 wala imputation: ',X_train['Age_99'].var())
print('Age Variance after -1 wala imputation: ', X_train['Age_minus1'].var())

print('Original Fare variable variance: ', X_train['Fare'].var())
print('Fare Variance after 999 wala imputation: ', X_train['Fare_999'].var())
print('Fare Variance after -1 wala imputation: ', X_train['Fare_minus1'].var())
```

```python
fig=plt.figure()
ax=fig.add_subplot(111)

# original variable distribution
X_train['Age'].plot(kind='kde',ax=ax)

# variable imputed with 99 wala imputatin
X_train['Age_99'].plot(kind='kde',ax=ax)

# Variable imputed with minus1 wala imputation
X_train['Age_minus1'].plot(kind='kde',ax=ax)

# add legends
lines,labels=ax.get_legend_handles_labels()
ax.legend(lines,labels,loc='best')

plt.show()
```

```python
fig = plt.figure()
ax = fig.add_subplot(111)

# original variable distribution
X_train['Fare'].plot(kind='kde', ax=ax)

# variable imputed with the 999 wala
X_train['Fare_999'].plot(kind='kde', ax=ax, color='red')

# variable imputed with the minux1 wala
X_train['Fare_minus1'].plot(kind='kde', ax=ax, color='green')

# add legends
lines, labels = ax.get_legend_handles_labels()
ax.legend(lines, labels, loc='best')
```

### Using Sklearn

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
imputer1=SimpleImputer(strategy='constant',fill_value=99)
imputer2=SimpleImputer(strategy='constant',fill_value=999)
```

```python
trf=ColumnTransformer([
    ('imputer1',imputer1,['Age']),
    ('imputer2',imputer2,['Fare'])
],remainder='passthrough')
```

```python
trf.fit(X_train)
```

```python
trf.named_transformers_['imputer1'].statistics_
```

```python
trf.named_transformers_['imputer2'].statistics_
```

```python
X_train=trf.fit_transform(X_train)
X_test=trf.fit_transform(X_test)
```
