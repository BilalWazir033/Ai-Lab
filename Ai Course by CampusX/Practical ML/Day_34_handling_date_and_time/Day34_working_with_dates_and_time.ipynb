```python
import numpy as np
import pandas as pd
```

```python
date=pd.read_csv('orders.csv')
time=pd.read_csv('messages.csv')
```

```python
date.head()
```

```python
date.info() # date is object should be date so
```

```python
# Converting to datetime datatype
date['date']=pd.to_datetime(date['date'])
```

```python
date.info()
```

### 1. Extract year

```python
date['date_year']=date['date'].dt.year
```

```python
date.head()
```

### 1. Extract month

```python
date['date_mon']=date['date'].dt.month
```

```python
date.head()
```

```python
# month name
date['date_mon_name']=date['date'].dt.month_name()
date.head()
```

### 1. Extract days


```python
date['date_day']=date['date'].dt.day
date.head()
```

```python
# day of week
date['date_dow']=date['date'].dt.day_of_week
date.head()
```

```python
# day of week name
date['date_dow_name']=date['date'].dt.day_name()
date.head()
```

```python
# is weekend?
date['date_is_weekend']=np.where(date['date_dow_name'].isin(['Sunday','Saturday']),1,0)
date.drop(columns=['product_id','city_id','orders']).head()
```

### Extract week of the year

```python
date['date_week']=date['date'].dt.isocalendar().week
date.drop(columns=['product_id','city_id','orders']).head()
```

### Extract Quarter

```python
date['quarter']=date['date'].dt.quarter
date.drop(columns=['product_id','city_id','orders']).head()
```

### Extract Semester

```python
date['semester']=np.where(date['quarter'].isin([1,2]),1,2)
date.drop(columns=['product_id','city_id','orders']).sample(5)
```

### Extract Time elapsed between dates

```python
import datetime
```

```python
today=datetime.datetime.today()
today
```

```python
today-date['date']
```

```python
(today-date['date']).dt.days
```

```python
# month passed
np.round((today - date['date']).dt.days / 30, 0) # here dt.days count total days, then divide by 30 
# to get months and round 0 means 2.4== 2
```
