```python
from sklearn.datasets import make_regression
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import plotly.express as px
```

```python
# making self dataset
X,y=make_regression(n_samples=100,n_features=2,n_informative=2,n_targets=1,noise=50) # sample total rows, n_features input cols, n_targets target cols
```

```python
# converting to dataframe
df=pd.DataFrame({'feature1':X[:,0],'feature2':X[:,1],'target':y})
```

```python
df.head()
```

```python
df.shape
```

```python
fig=px.scatter_3d(df,x='feature1',y='feature2',z='target')
fig.show()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=3)
```

```python
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
```

```python
lr.fit(X_train,y_train)
```

```python
y_predict=lr.predict(X_test)
```

```python
y_predict
```

```python
from sklearn.metrics import accuracy_score,mean_absolute_error,mean_squared_error,r2_score
```

```python
mae=mean_absolute_error(y_test,y_predict)
mse=mean_squared_error(y_test,y_predict)
r2=r2_score(y_test,y_predict)
```

```python
print(f'MAE: {mae}, MSE: {mse}, R2: {r2}')
```

```python
# Import NumPy for numerical operations
import numpy as np

# Import Plotly Express for creating the 3D scatter plot
import plotly.express as px

# Import Plotly Graph Objects for creating the regression surface/plane
import plotly.graph_objects as go


# Create 10 equally spaced values between -5 and 5 for feature 1
x = np.linspace(-5, 5, 10)

# Create 10 equally spaced values between -5 and 5 for feature 2
y = np.linspace(-5, 5, 10)


# Create a grid of x and y values
# This generates all possible combinations of x and y
xGrid, yGrid = np.meshgrid(x, y)


# Convert the grid values into a list of (x, y) pairs
# ravel() converts the 2D arrays into 1D arrays
# vstack() combines x and y values
# .T transposes the result so each row becomes [x, y]
final = np.vstack((
    xGrid.ravel(),
    yGrid.ravel()
)).T


# Use the trained Linear Regression model to predict
# the target value for every (x, y) combination
z_final = lr.predict(final)


# Convert the predicted values into a 10 x 10 matrix
# so they can be used to create the 3D surface
z_final = z_final.reshape(10, 10)


# Store the predicted z values in variable z
z = z_final


# Create a 3D scatter plot using the original dataset
# feature1 → X-axis
# feature2 → Y-axis
# target   → Z-axis
fig = px.scatter_3d(
    df,
    x='feature1',
    y='feature2',
    z='target'
)


# Add the Linear Regression prediction surface/plane
# x and y represent the two input features
# z represents the predicted target values
fig.add_trace(
    go.Surface(
        x=x,
        y=y,
        z=z
    )
)


# Display the final 3D graph
fig.show()
```
