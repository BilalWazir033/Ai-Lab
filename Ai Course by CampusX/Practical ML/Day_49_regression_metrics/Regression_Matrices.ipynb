```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df=pd.read_csv("placement.csv")
```

```python
df.head()
```

```python
df.shape
```

```python
plt.scatter(df['cgpa'],df['package'])
plt.xlabel('CGPA')
plt.ylabel('Package') 
plt.show()
```

```python
X=df.iloc[:,0:1]
X
```

```python
y=df.iloc[:,-1]
y
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=2)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr=LinearRegression()
```

```python
lr.fit(X_train,y_train)
```

```python
y_pred=lr.predict(X_test)
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score,accuracy_score
```

```python
plt.scatter(df['cgpa'],df['package'])
plt.plot(X_train,lr.predict(X_train),color='red')
plt.xlabel('CGPA')
plt.ylabel('Package')
plt.show()
```

```python
MAE=mean_absolute_error(y_test,y_pred)
MSE=mean_squared_error(y_test,y_pred)
RSME=np.sqrt(MSE)
R2=r2_score(y_test,y_pred)
```

```python
print("Mean Absolute Error:",MAE)
print("Mean Squared Error:",MSE)
print("Root Mean Squared Error:",RSME)
print("R2 Score:",R2)     
```

# Adjusted R2 Score

```python
X_test.shape
```

```python
1-(1-R2)*(len(y_test)-1)/(len(y_test)-X_test.shape[1]-1)
```

### Giving an Irrelevant feature

```python
df1=df.copy()
df1.head()
```

```python
df1['Random_Feature']=np.random.random(200)
df1=df1[['cgpa','Random_Feature','package']]
df1.head()
```

```python
plt.scatter(df1['package'],df1['Random_Feature'])
plt.xlabel('Package')
plt.ylabel('Random Feature')
plt.title('Scatter Plot of Package vs Random Feature')
plt.show()
```

```python
X = df1.iloc[:,:2]
y = df1.iloc[:, -1]
```

```python
X
```

```python
y
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=2)
```

```python
lr.fit(X_train,y_train)
```

```python
y_pred=lr.predict(X_test)
```

```python
r2=r2_score(y_test,y_pred)
print("R2 Score:", r2)
```

```python
# Adjusted R2 Score decreased
R2_adj = 1 - (1 - r2) * (len(y_test) - 1) / (len(y_test) - X_test.shape[1] - 1)
print("Adjusted R2 Score:", R2_adj)
```
