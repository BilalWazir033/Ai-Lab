## 2. Bar Plot (Numerical - Categorical)

```python
titanic=pd.read_csv('train.csv')
```

```python
titanic.head()
```

```python
sns.barplot(y=titanic['Age'],x=titanic['Embarked'],palette='Set1')
```

```python
sns.barplot(y=titanic['Age'],x=titanic['Embarked'],hue=titanic['Sex'],palette='Set1')
```

```python
sns.catplot(data=titanic, x='Pclass', y='Age', hue='Sex', col='Embarked', kind='bar')
```

## 3. Box Plot (Numerical - Categorical)

```python
sns.boxplot(data=titanic, x='Sex',y='Age')
```

```python
sns.boxplot(data=titanic, x='Sex',y='Age',hue='Embarked')
```

### 4. Distplot (Numerical - Categorical)

```python
sns.displot(titanic[titanic['Survived']==0]['Age'],stat='density') # PDF of people not survived
```

```python
sns.displot(titanic[titanic['Survived']==1]['Age'],stat='density') # people survived
```

```python
sns.displot(titanic[titanic['Survived']==0]['Age'],stat='density',kde=True) 
sns.displot(titanic[titanic['Survived']==1]['Age'],stat='density',kde=True)
```

```python
sns.kdeplot(data=titanic, x='Age', hue='Survived', fill=True) # easy approach
```

```python
sns.displot(data=titanic, x='Age', hue='Survived', col='Sex', kde=True)
```

```python
sns.displot(data=titanic, x='Age', hue='Survived', col='Sex', kind='kde', fill=False)
```

### 5. HeatMap (Categorical - Categorical)

```python
sns.heatmap(pd.crosstab(titanic['Pclass'],titanic['Survived']))
```

### 8. Lineplot (Numerical - Numerical)

```python
flights=pd.read_csv('flights.csv',usecols=['year','month','distance','day'])
flights.sample(20)
```

```python
new=flights.groupby('month').sum().reset_index()
new
```

```python
sns.lineplot(x=new['month'],y=new['distance'])
```

```python
flights.pivot_table(values='distance',index='day',columns='month')
```

```python
sns.clustermap(flights.pivot_table(values='distance',index='day',columns='month'))
```
