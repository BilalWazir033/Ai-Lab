```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv('titanic.csv')
```

```python
fig=df['number'].value_counts().plot.bar()
fig.set_title('Passengers travelling with')
```

```python
# extract numerical part from number 
df['number_num']=pd.to_numeric(df['number'],errors='coerce',downcast='integer')

# errors='coerce'
# If a value cannot be converted into a number:it becomes NaN

# downcast='integer'
# Reduces memory usage by converting numbers to the smallest suitable integer type.
```

```python
# extract categorical part from number column
df['number_cat']=np.where(df['number_num'].isnull(),df['number'],np.nan)
```

```python
df['Cabin_num']=df['Cabin'].str.extract('(\d+)') # capture numerical part
df['Cabin_cat']=df['Cabin'].str[0] # capture the first letter

# Regex Meaning
# (\d+)
# \d → any digit (0–9)
# + → one or more digits
# () → capture the matched part
```

```python
# extract the last bit of ticket as number
df['ticket_num']=df['Ticket'].apply(lambda s: s.split()[-1])
df['ticket_num']=pd.to_numeric(df['ticket_num'],errors='coerce',downcast='integer')


# extract the first part of ticket as category
df['ticket_cat']=df['Ticket'].apply(lambda s: s.split()[0])
df['ticket_cat']=np.where(df['ticket_cat'].str.isdigit(),np.nan,df['ticket_cat'])
```
