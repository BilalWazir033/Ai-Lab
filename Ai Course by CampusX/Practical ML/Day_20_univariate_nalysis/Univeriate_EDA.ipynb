```python
# to find Q1
q1=df['Age'].quantile(0.25)
q1
```

```python
q3=df['Age'].quantile(0.75)
q3
```

```python
IQR=q3-q1
IQR
```

```python
mini=q1-1.5*IQR
mini
```

```python
maxi=q3+1.5*IQR
maxi
```

```python
q2=df['Age'].quantile(0.50)
q2
```

```python
outliers=df[(df['Age']<mini) | (df['Age']>maxi)]
outliers
```
