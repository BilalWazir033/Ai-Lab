```python
# Choice A: For text columns, fill blanks with the word "Unknown
df[['ship_date', 'carrier', 'priority']] = df[['ship_date', 'carrier', 'priority']].fillna('Unknown')
```

```python
# Choice B: For number columns, fill blanks with the median (the middle number)
num_cols = df.select_dtypes(include='number').columns
df[num_cols] = df[num_cols].fillna(df[num_cols].median())
```

```python
# Check for duplicates and drop them
df.duplicated().sum()
df=df.drop_duplicates()
```

```python
# Make text lowercase and strip hidden spaces
df[['carrier', 'priority']] = df[['carrier', 'priority']].apply(lambda x: x.str.lower().str.strip())
```

```python
# Example: Keep only rows where the Price is greater than 0
df = df[df['shipping_cost_usd'] > 0]
```

```python
# Convert a text date into a real Python Datetime
df['ship_date'] = pd.to_datetime(df['ship_date'], errors='coerce')

```

```python
# Create a brand new column that just contains the Month (1 through 12)
df['ship_month'] = df['ship_date'].dt.month
```

```python
# Create a new column by dividing or multiplying two others
df['cost_per_kg'] = df['shipping_cost_usd'] / df['weight_kg']
```

```python
# Convert text categories into separate columns of 1s and 0s (One-Hot Encoding)
df_encoded = pd.get_dummies(df, columns=['carrier', 'priority'], drop_first=True)
```
