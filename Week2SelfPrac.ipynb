### np.linspace(start, stop, num=50, ...)
- start → the first number in the sequence=
- stop → the last number in the sequence (included by default)
- num → how many numbers you want in total (default is 50)
#### It creates evenly spaced numbers between start and stop (means that space will be equal in subsequent number)

```python
arr=np.linspace(2,8,10)
print(arr)
```

## Copy vs View

```python
arr=np.array([23,5,64,7])
a=arr.copy() # makes a completely independent copy
b=arr.view() # makes a new view pointing to the same data
a[0]=100 
print(arr) # original array stays the same
print(a)  # copy changed
b[0]=100
print(arr) # original array also changes
print(b) # view changed
```
